```python



# 1. Creating a set
s = {1, 2, 3, "hello", 1, 2, 3}
print(s)

# 2. Empty set
A = {}
print(type(A))       # dict

A = set()
print(type(A))       # set

# 3. Union, intersection, difference, symmetric difference
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)         # Union
print(A & B)         # Intersection
print(A - B)         # Difference
print(A ^ B)         # Symmetric difference

# 4. Set from lists
L = [12, 33, 44]
L1 = [33, 45, 67]

A = set(L)
B = set(L1)

print(A & B)

# 5. Set with different data types
s = {1, "hello", 22.3, (1, 2, 3)}
print(s)

# 6. Adding elements
s.add(23)
print(s)

# 7. Update multiple elements
s.update((1, 2), 3, 44, 5)
print(s)

# 8. Printing two sets
A = {1, 2, 3}
B = {3, 4, 5}
print(A, B)

# 9. Union using method
print(A.union({5, 6, 7, 8}))

# 10. Intersection
C = {2, 3, 4, 5}
print(A.intersection(C))

# 11. Symmetric difference update
A = {1, 2, 3}
A.symmetric_difference_update(C)
print(A)

# 12. Set from a number
a = 333
s = str(a)
l = set(s)
print(l)

# 13. Frozenset
s = frozenset((1, 2, 3))
print(s)

# 14. Set methods
s = set()
s.update((1, 2, 3, 4, 5))

s.add(10)
print(s)

s.remove(10)
print(s)

s.discard(100)
print(s)

s.pop()
print(s)

s.clear()
print(s)

# 15. Set difference using range
n = 5
numbers = set(range(1, n + 1))
L = [1, 1, 2, 2, 5, 3]
L2 = set(L)
print(numbers - L2)
```
