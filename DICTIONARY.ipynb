```python


# 1. Creating a dictionary
d = {
    "name": "sai",
    "age": 20,
    "grade": 9.3,
    "height": 6.0,
    "NAME": "sri"
}
print(d)

# 2. Empty dictionary
d = {}
print(type(d))

# 3. Dictionary using dict()
d1 = dict(a=70, n=34)
print(d1)

# 4. Accessing dictionary
d = {
    "name": "sai",
    "age": 20,
    "grade": 9.3,
    "height": 6.0
}

# 5. Printing keys
for i in d:
    print(i)

# 6. Printing values
for i in d:
    print(d[i])

# 7. Printing key and value
for i in d:
    print(i, ":", d[i])

# 8. Dictionary keys
x = d.keys()
print(x)
print(type(x))
print(list(x))

# 9. Dictionary values
y = d.values()
print(y)
print(list(y))

# 10. Dictionary items
z = d.items()
print(z)

# 11. Converting items into list
z = list(z)
print(z)

# 12. Printing items using loop
for i in d.items():
    print(i)

# 13. Adding a new key and value
d["perc"] = 89
print(d)

# 14. Updating dictionary
d.update({"perc2": 78})
print(d)

# 15. Removing an item using pop()
print(d.pop("name"))

# 16. Printing dictionary after pop
print(d)

# 17. Dictionary comprehension
d2 = {i: i**2 for i in range(1, 5)}
print(d2)

# 18. Creating dictionary using zip()
k = ["sno", "age", "reg"]
v = [1, 22, 3445]

d = {i: j for i, j in zip(k, v)}
print(d)

# 19. Dictionary using zip()
d = dict(zip(k, v))
print(d)
```
