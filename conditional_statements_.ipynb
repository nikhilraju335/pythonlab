# **program on match**

```python
num=int(input("enter number"))
match num:
  case 1:
    print(1)
  case 2:
    print(2)
  case 3:
    print(3)
  case _:
    print("invalid")
```

```python
n=int(input("enter a number:"))
x=int(input("enter a to multiply upto"))
for i in range(0,x+1,1):
  r=n * i
  print(f"{n} * {i} = {r}")
```
